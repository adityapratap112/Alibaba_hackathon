# 🔍 Crypto Sentiment Analysis - Web Scraping Notebook
## Deep Sentiment Analysis from Major Crypto News Sources

**Sources:** CoinDesk | CoinTelegraph | BeInCrypto | U.Today | The Block

**Tools:** Tavily API | BeautifulSoup | Selenium | **🤖 DeepSeek AI** | VADER & TextBlob

This notebook provides comprehensive sentiment analysis by scraping news from major crypto sources and analyzing them using multiple sentiment analysis techniques.

```python
# Import the sentiment scraper functions
import sys
sys.path.append('/Users/aditya/PycharmProjects/PAlibaba_hackathon')

# Import all functions from our script
from crypto_sentiment_scraper import (
    search_with_tavily,
    scrape_with_bs,
    analyze_sentiment_vader,
    analyze_sentiment_textblob,
    analyze_sentiment_ai,
    NEWS_SOURCES,
    tavily_key,
    openrouter_key
)

import time
from datetime import datetime

print("✅ Sentiment analysis functions imported")
print(f"📰 {len(NEWS_SOURCES)} news sources configured")
print(f"🔑 Tavily: {'✅' if tavily_key else '❌'}")
print(f"🔑 OpenRouter: {'✅' if openrouter_key else '❌'}")

# DeepSeek AI Status Check
print(f"\n🤖 **DeepSeek AI Integration Status:**")
if openrouter_key:
    print("   ✅ DeepSeek AI: READY FOR ADVANCED ANALYSIS")
    print("   • Professional market sentiment analysis")
    print("   • Trading recommendations with confidence levels")
    print("   • Risk assessment and price predictions")
    print("   • Market context and sentiment drivers")
else:
    print("   ⚠️  DeepSeek AI: API key needed")
    print("   • Add OPENROUTER_API_KEY to .env for AI analysis")

print(f"📅 Current time: {datetime.now()}")
```

```python
# User input - Enter cryptocurrency name
coin_name = input("🪙 Enter cryptocurrency (e.g., Bitcoin, Ethereum, Solana): ").strip()

if not coin_name:
    coin_name = "Bitcoin"
    print(f"Using default: {coin_name}")

print(f"\n🔍 Starting comprehensive sentiment analysis for: {coin_name}")
print("=" * 70)

all_articles = []
```

```python
# Step 1: Search with Tavily API
print("🔍 STEP 1: Tavily Search")
print("-" * 30)

tavily_articles = search_with_tavily(coin_name, max_results=8)
all_articles.extend(tavily_articles)

if tavily_articles:
    print("\nTop Tavily results:")
    for i, article in enumerate(tavily_articles[:3], 1):
        print(f"{i}. {article['title'][:70]}...")
        print(f"   Source: {article['source']}")
        print(f"   URL: {article['url'][:60]}...")
        print()
```

```python
# Step 2: Scrape with BeautifulSoup
print("🍲 STEP 2: BeautifulSoup Scraping")
print("-" * 30)

bs_articles = []
for source_name in NEWS_SOURCES.keys():
    articles = scrape_with_bs(coin_name, source_name)
    bs_articles.extend(articles)
    time.sleep(1)  # Be respectful to servers

all_articles.extend(bs_articles)
print(f"\n📊 BeautifulSoup total: {len(bs_articles)} articles")
```

```python
# Step 3: Process and filter articles
print("\n🔄 STEP 3: Processing Articles")
print("-" * 30)

# Remove duplicates
unique_articles = []
seen_titles = set()

for article in all_articles:
    title = article.get('title', '').lower().strip()
    if title and len(title) > 10:
        title_key = title[:40]  # First 40 chars for dedup
        if title_key not in seen_titles:
            seen_titles.add(title_key)
            unique_articles.append(article)

# Filter for relevance
relevant_articles = []
coin_keywords = [coin_name.lower(), coin_name.lower()[:3]]  # e.g., bitcoin, btc

for article in unique_articles:
    text = (article.get('title', '') + ' ' + article.get('content', '')).lower()
    
    if any(keyword in text for keyword in coin_keywords) and len(article.get('content', '')) > 80:
        relevant_articles.append(article)

print(f"📊 Processing Results:")
print(f"   • Total scraped: {len(all_articles)}")
print(f"   • After deduplication: {len(unique_articles)}")
print(f"   • Relevant to {coin_name}: {len(relevant_articles)}")

# Show source breakdown
sources = {}
for article in relevant_articles:
    source = article.get('source', 'Unknown')
    sources[source] = sources.get(source, 0) + 1

print(f"\n📰 Articles by source:")
for source, count in sorted(sources.items(), key=lambda x: x[1], reverse=True):
    print(f"   • {source}: {count}")
```

```python
# Step 4: Perform sentiment analysis
print("\n🧠 STEP 4: Sentiment Analysis")
print("-" * 30)

if not relevant_articles:
    print("❌ No relevant articles found for analysis")
else:
    print(f"Analyzing {len(relevant_articles)} articles...\n")
    
    all_sentiments = []
    
    for i, article in enumerate(relevant_articles[:8], 1):  # Limit for notebook
        print(f"📄 Article {i}: {article['title'][:60]}...")
        print(f"   Source: {article['source']}")
        
        # Combine title and content for analysis
        full_text = article['title'] + ' ' + article['content']
        
        # VADER analysis
        vader_result = analyze_sentiment_vader(full_text)
        print(f"   🎯 VADER: {vader_result['sentiment']} ({vader_result['compound']:.2f})")
        
        # TextBlob analysis
        textblob_result = analyze_sentiment_textblob(full_text)
        print(f"   📊 TextBlob: {textblob_result['sentiment']} ({textblob_result['polarity']:.2f})")
        
        # AI analysis (for first 3 articles)
        if i <= 3 and openrouter_key:
            ai_result = analyze_sentiment_ai(full_text, coin_name)
            if ai_result.get('success'):
                print(f"   🤖 AI Analysis: {ai_result['analysis'][:80]}...")
            else:
                print(f"   🤖 AI Analysis: {ai_result.get('error', 'Failed')}")
        
        # Store results
        sentiment_data = {
            'article': article,
            'vader': vader_result,
            'textblob': textblob_result
        }
        
        if i <= 3 and openrouter_key:
            sentiment_data['ai'] = ai_result
        
        all_sentiments.append(sentiment_data)
        print()
```

```python
# Step 4.5: 🤖 Enhanced DeepSeek AI Analysis
print("\n🤖 STEP 4.5: Enhanced DeepSeek AI Market Analysis")
print("=" * 60)

if all_sentiments and openrouter_key:
    print(f"🧠 Generating comprehensive AI analysis for {coin_name}...")

    # Prepare comprehensive data for AI analysis
    all_titles = [s['article']['title'] for s in all_sentiments]
    all_content = [s['article']['content'][:300] for s in all_sentiments]  # Limit content
    vader_scores = [s['vader']['compound'] for s in all_sentiments]
    textblob_scores = [s['textblob']['polarity'] for s in all_sentiments]

    # Create comprehensive prompt for DeepSeek
    comprehensive_prompt = f"""As a cryptocurrency market analyst, provide a comprehensive sentiment analysis for {coin_name} based on the following data:

RECENT NEWS HEADLINES:
{chr(10).join([f"• {title}" for title in all_titles[:5]])}

SENTIMENT SCORES:
• VADER Average: {sum(vader_scores)/len(vader_scores):.3f}
• TextBlob Average: {sum(textblob_scores)/len(textblob_scores):.3f}
• Articles Analyzed: {len(all_sentiments)}

Please provide:
1. **Overall Market Sentiment**: Bullish/Bearish/Neutral with confidence level (1-10)
2. **Key Sentiment Drivers**: What themes are driving the sentiment?
3. **Price Impact Prediction**: Short-term (1-7 days) price movement expectation
4. **Risk Assessment**: Potential risks or opportunities
5. **Trading Recommendation**: Buy/Hold/Sell with reasoning
6. **Market Context**: How does this compare to broader crypto market sentiment?

Format your response clearly with emojis and be specific about your confidence levels."""

    try:
        from openai import OpenAI
        client = OpenAI(
            base_url="https://openrouter.ai/api/v1",
            api_key=openrouter_key
        )

        print("🔄 Analyzing with DeepSeek AI...")

        response = client.chat.completions.create(
            model="deepseek/deepseek-chat-v3.1:free",
            messages=[
                {"role": "system", "content": "You are a professional cryptocurrency market analyst with expertise in sentiment analysis, technical analysis, and market psychology. Provide detailed, actionable insights."},
                {"role": "user", "content": comprehensive_prompt}
            ],
            max_tokens=800,
            temperature=0.3
        )

        ai_analysis = response.choices[0].message.content.strip()

        print("\n🤖 **DeepSeek AI Market Analysis:**")
        print("=" * 60)
        print(ai_analysis)

        # Store AI analysis for later use
        deepseek_analysis = {
            'analysis': ai_analysis,
            'timestamp': datetime.now(),
            'model': 'deepseek-chat-v3.1',
            'confidence': 'High',
            'data_points': len(all_sentiments)
        }

    except Exception as e:
        print(f"❌ DeepSeek AI analysis failed: {e}")
        deepseek_analysis = None

elif not openrouter_key:
    print("⚠️  DeepSeek AI analysis skipped - OpenRouter API key not found")
    deepseek_analysis = None
else:
    print("⚠️  DeepSeek AI analysis skipped - No sentiment data available")
    deepseek_analysis = None
```

```python
# Step 5: Generate comprehensive summary
print("📊 STEP 5: Comprehensive Sentiment Summary")
print("=" * 60)

if all_sentiments:
    # Calculate overall sentiment scores
    vader_scores = [s['vader']['compound'] for s in all_sentiments]
    textblob_scores = [s['textblob']['polarity'] for s in all_sentiments]
    
    vader_avg = sum(vader_scores) / len(vader_scores)
    textblob_avg = sum(textblob_scores) / len(textblob_scores)
    
    # Count sentiment categories
    vader_sentiments = [s['vader']['sentiment'] for s in all_sentiments]
    textblob_sentiments = [s['textblob']['sentiment'] for s in all_sentiments]
    
    vader_counts = {s: vader_sentiments.count(s) for s in ['Positive', 'Negative', 'Neutral']}
    textblob_counts = {s: textblob_sentiments.count(s) for s in ['Positive', 'Negative', 'Neutral']}
    
    print(f"🪙 **{coin_name} Sentiment Analysis Results**\n")
    
    print(f"📈 **Overall Sentiment Scores:**")
    print(f"   • VADER Average: {vader_avg:.3f} ({'Positive' if vader_avg > 0.05 else 'Negative' if vader_avg < -0.05 else 'Neutral'})")
    print(f"   • TextBlob Average: {textblob_avg:.3f} ({'Positive' if textblob_avg > 0.1 else 'Negative' if textblob_avg < -0.1 else 'Neutral'})")
    
    print(f"\n📊 **Sentiment Distribution:**")
    print(f"   VADER: {vader_counts['Positive']} Positive, {vader_counts['Negative']} Negative, {vader_counts['Neutral']} Neutral")
    print(f"   TextBlob: {textblob_counts['Positive']} Positive, {textblob_counts['Negative']} Negative, {textblob_counts['Neutral']} Neutral")
    
    # Determine overall market sentiment
    if vader_avg > 0.1 and textblob_avg > 0.1:
        overall_sentiment = "🟢 **BULLISH** - Strong positive sentiment"
    elif vader_avg < -0.1 and textblob_avg < -0.1:
        overall_sentiment = "🔴 **BEARISH** - Strong negative sentiment"
    elif vader_avg > 0.05 or textblob_avg > 0.05:
        overall_sentiment = "🟡 **CAUTIOUSLY OPTIMISTIC** - Mild positive sentiment"
    elif vader_avg < -0.05 or textblob_avg < -0.05:
        overall_sentiment = "🟠 **CAUTIOUSLY PESSIMISTIC** - Mild negative sentiment"
    else:
        overall_sentiment = "⚪ **NEUTRAL** - Mixed or neutral sentiment"
    
    print(f"\n🎯 **Overall Market Sentiment for {coin_name}:**")
    print(f"   {overall_sentiment}")
    
    print(f"\n📰 **Source Breakdown:**")
    source_sentiments = {}
    for sentiment_data in all_sentiments:
        source = sentiment_data['article']['source']
        vader_sentiment = sentiment_data['vader']['sentiment']
        if source not in source_sentiments:
            source_sentiments[source] = []
        source_sentiments[source].append(vader_sentiment)
    
    for source, sentiments in source_sentiments.items():
        pos = sentiments.count('Positive')
        neg = sentiments.count('Negative')
        neu = sentiments.count('Neutral')
        print(f"   • {source}: {pos}+ {neg}- {neu}○")
    
    print(f"\n🔍 **Key Insights:**")
    print(f"   • Analyzed {len(all_sentiments)} articles from {len(source_sentiments)} sources")
    print(f"   • Data collected: {datetime.now().strftime('%Y-%m-%d %H:%M')}")
    print(f"   • Sentiment range: {min(vader_scores):.3f} to {max(vader_scores):.3f}")

    # Add DeepSeek AI summary if available
    if 'deepseek_analysis' in locals() and deepseek_analysis:
        print(f"\n🤖 **DeepSeek AI Summary:**")
        print(f"   • AI Model: {deepseek_analysis['model']}")
        print(f"   • Analysis Confidence: {deepseek_analysis['confidence']}")
        print(f"   • Data Points Used: {deepseek_analysis['data_points']}")
        print(f"   • AI Analysis Available: ✅ Comprehensive market insights generated")
    
    # Show most positive and negative articles
    if len(all_sentiments) > 1:
        most_positive = max(all_sentiments, key=lambda x: x['vader']['compound'])
        most_negative = min(all_sentiments, key=lambda x: x['vader']['compound'])
        
        print(f"\n📈 **Most Positive Article:**")
        print(f"   {most_positive['article']['title'][:80]}...")
        print(f"   Score: {most_positive['vader']['compound']:.3f} | Source: {most_positive['article']['source']}")
        
        print(f"\n📉 **Most Negative Article:**")
        print(f"   {most_negative['article']['title'][:80]}...")
        print(f"   Score: {most_negative['vader']['compound']:.3f} | Source: {most_negative['article']['source']}")

else:
    print("❌ No sentiment analysis data available")

print(f"\n🎉 **Advanced Sentiment Analysis Complete for {coin_name}!**")

# Final DeepSeek-Enhanced Summary
if 'deepseek_analysis' in locals() and deepseek_analysis:
    print(f"\n🚀 **Enhanced with DeepSeek AI Analysis:**")
    print(f"   ✅ Traditional sentiment analysis (VADER + TextBlob)")
    print(f"   ✅ Web scraping from major crypto news sources")
    print(f"   ✅ AI-powered market analysis and predictions")
    print(f"   ✅ Professional trading recommendations")
    print(f"   ✅ Risk assessment and market context")
    print(f"\n💡 **This analysis combines multiple data sources with advanced AI to provide**")
    print(f"   **the most comprehensive crypto sentiment insights available!**")
else:
    print(f"\n📊 **Comprehensive Multi-Method Analysis Complete:**")
    print(f"   ✅ VADER sentiment analysis")
    print(f"   ✅ TextBlob sentiment analysis")
    print(f"   ✅ Web scraping from major sources")
    print(f"   ✅ Statistical sentiment aggregation")
    print(f"\n💡 **Add OpenRouter API key to enable DeepSeek AI analysis for even deeper insights!**")
```
