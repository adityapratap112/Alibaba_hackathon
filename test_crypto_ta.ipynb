## 5. Test Data Fetching

```python
# Test data fetching for multiple symbols
print("🧪 Testing Data Fetching...")
print("=" * 40)

test_results = {}

for symbol in TEST_SYMBOLS:
    print(f"\n📊 Testing {symbol}...")
    
    try:
        # Fetch data
        raw_data = get_crypto_data(symbol, period=TEST_PERIOD, interval=TEST_INTERVAL)
        
        if raw_data is not None and not raw_data.empty:
            # Validate data
            validated_data = validate_data(raw_data)
            
            if validated_data is not None and not validated_data.empty:
                test_results[symbol] = {
                    'status': 'SUCCESS',
                    'data': validated_data,
                    'records': len(validated_data),
                    'date_range': f"{validated_data.index[0].strftime('%Y-%m-%d')} to {validated_data.index[-1].strftime('%Y-%m-%d')}",
                    'current_price': validated_data['Close'].iloc[-1]
                }
                print(f"   ✅ SUCCESS: {len(validated_data)} records")
                print(f"   💰 Current price: ${validated_data['Close'].iloc[-1]:,.2f}")
            else:
                test_results[symbol] = {'status': 'VALIDATION_FAILED'}
                print(f"   ❌ FAILED: Data validation failed")
        else:
            test_results[symbol] = {'status': 'NO_DATA'}
            print(f"   ❌ FAILED: No data retrieved")
            
    except Exception as e:
        test_results[symbol] = {'status': 'ERROR', 'error': str(e)}
        print(f"   ❌ ERROR: {e}")

# Summary
print("\n📋 Data Fetching Test Summary:")
print("=" * 35)
successful_tests = sum(1 for result in test_results.values() if result['status'] == 'SUCCESS')
print(f"✅ Successful: {successful_tests}/{len(TEST_SYMBOLS)}")

for symbol, result in test_results.items():
    status_emoji = "✅" if result['status'] == 'SUCCESS' else "❌"
    print(f"{status_emoji} {symbol}: {result['status']}")
```

## 6. Test Technical Analysis

```python
# Test technical analysis on successful data
print("🧪 Testing Technical Analysis...")
print("=" * 40)

analysis_results = {}

for symbol, result in test_results.items():
    if result['status'] == 'SUCCESS':
        print(f"\n📊 Analyzing {symbol}...")
        
        try:
            # Create analyzer
            analyzer = TechnicalAnalyzer(result['data'])
            
            # Calculate indicators
            analyzed_data = analyzer.calculate_all_indicators()
            
            # Get latest values for analysis
            latest = analyzed_data.iloc[-1]
            
            analysis_results[symbol] = {
                'status': 'SUCCESS',
                'data': analyzed_data,
                'current_price': latest['Close'],
                'rsi': latest.get('RSI', 'N/A'),
                'macd_signal': 'Bullish' if latest.get('MACD', 0) > latest.get('MACD_Signal', 0) else 'Bearish',
                'sma20_trend': 'Above' if latest['Close'] > latest.get('SMA_20', latest['Close']) else 'Below',
                'bb_position': 'Normal'
            }
            
            # Bollinger Bands position
            if 'BB_Upper' in analyzed_data.columns and 'BB_Lower' in analyzed_data.columns:
                if latest['Close'] > latest['BB_Upper']:
                    analysis_results[symbol]['bb_position'] = 'Above Upper Band'
                elif latest['Close'] < latest['BB_Lower']:
                    analysis_results[symbol]['bb_position'] = 'Below Lower Band'
                else:
                    analysis_results[symbol]['bb_position'] = 'Within Bands'
            
            print(f"   ✅ SUCCESS: All indicators calculated")
            print(f"   💰 Price: ${latest['Close']:,.2f}")
            print(f"   ⚡ RSI: {latest.get('RSI', 'N/A'):.1f}" if isinstance(latest.get('RSI'), (int, float)) else f"   ⚡ RSI: N/A")
            print(f"   📊 MACD: {analysis_results[symbol]['macd_signal']}")
            
        except Exception as e:
            analysis_results[symbol] = {'status': 'ERROR', 'error': str(e)}
            print(f"   ❌ ERROR: {e}")
    else:
        print(f"\n⏭️ Skipping {symbol} (no valid data)")

# Summary
print("\n📋 Technical Analysis Test Summary:")
print("=" * 40)
successful_analysis = sum(1 for result in analysis_results.values() if result['status'] == 'SUCCESS')
print(f"✅ Successful analyses: {successful_analysis}")

for symbol, result in analysis_results.items():
    if result['status'] == 'SUCCESS':
        status_emoji = "✅"
        print(f"{status_emoji} {symbol}: Technical analysis complete")
    else:
        print(f"❌ {symbol}: {result['status']}")
```

## 7. Create Test Visualizations

```python
# Create visualizations for successful analyses
print("🎨 Creating Test Visualizations...")
print("=" * 40)

for symbol, result in analysis_results.items():
    if result['status'] == 'SUCCESS':
        print(f"\n📊 Creating charts for {symbol}...")
        
        try:
            analyzed_data = result['data']
            
            # Create comprehensive matplotlib charts
            fig, axes = plt.subplots(4, 1, figsize=(15, 16))
            fig.suptitle(f'{symbol} Technical Analysis Test Results', fontsize=16, fontweight='bold')
            
            # 1. Price chart with moving averages
            axes[0].plot(analyzed_data.index, analyzed_data['Close'], label='Close Price', linewidth=2, color='blue')
            if 'SMA_20' in analyzed_data.columns:
                axes[0].plot(analyzed_data.index, analyzed_data['SMA_20'], label='SMA(20)', alpha=0.8, color='orange')
            if 'SMA_50' in analyzed_data.columns:
                axes[0].plot(analyzed_data.index, analyzed_data['SMA_50'], label='SMA(50)', alpha=0.8, color='red')
            if 'BB_Upper' in analyzed_data.columns:
                axes[0].plot(analyzed_data.index, analyzed_data['BB_Upper'], label='BB Upper', alpha=0.5, color='gray')
                axes[0].plot(analyzed_data.index, analyzed_data['BB_Lower'], label='BB Lower', alpha=0.5, color='gray')
                axes[0].fill_between(analyzed_data.index, analyzed_data['BB_Upper'], analyzed_data['BB_Lower'], alpha=0.1, color='gray')
            axes[0].set_title(f'{symbol} Price Chart with Technical Indicators', fontsize=12, fontweight='bold')
            axes[0].set_ylabel('Price ($)')
            axes[0].legend()
            axes[0].grid(True, alpha=0.3)
            
            # 2. RSI chart
            if 'RSI' in analyzed_data.columns:
                axes[1].plot(analyzed_data.index, analyzed_data['RSI'], label='RSI', color='purple', linewidth=2)
                axes[1].axhline(y=70, color='red', linestyle='--', alpha=0.7, label='Overbought (70)')
                axes[1].axhline(y=30, color='green', linestyle='--', alpha=0.7, label='Oversold (30)')
                axes[1].axhline(y=50, color='black', linestyle='-', alpha=0.3)
                axes[1].set_title('RSI (Relative Strength Index)', fontsize=12, fontweight='bold')
                axes[1].set_ylabel('RSI')
                axes[1].set_ylim(0, 100)
                axes[1].legend()
                axes[1].grid(True, alpha=0.3)
            
            # 3. MACD chart
            if 'MACD' in analyzed_data.columns:
                axes[2].plot(analyzed_data.index, analyzed_data['MACD'], label='MACD', color='blue', linewidth=2)
                if 'MACD_Signal' in analyzed_data.columns:
                    axes[2].plot(analyzed_data.index, analyzed_data['MACD_Signal'], label='Signal', color='red', linewidth=2)
                if 'MACD_Histogram' in analyzed_data.columns:
                    colors = ['green' if val >= 0 else 'red' for val in analyzed_data['MACD_Histogram']]
                    axes[2].bar(analyzed_data.index, analyzed_data['MACD_Histogram'], color=colors, alpha=0.6, label='Histogram')
                axes[2].axhline(y=0, color='black', linestyle='-', alpha=0.3)
                axes[2].set_title('MACD (Moving Average Convergence Divergence)', fontsize=12, fontweight='bold')
                axes[2].set_ylabel('MACD')
                axes[2].legend()
                axes[2].grid(True, alpha=0.3)
            
            # 4. Volume chart
            axes[3].bar(analyzed_data.index, analyzed_data['Volume'], alpha=0.6, color='lightblue', label='Volume')
            if 'Volume_SMA' in analyzed_data.columns:
                axes[3].plot(analyzed_data.index, analyzed_data['Volume_SMA'], color='red', linewidth=2, label='Volume SMA')
            axes[3].set_title('Trading Volume', fontsize=12, fontweight='bold')
            axes[3].set_ylabel('Volume')
            axes[3].set_xlabel('Date')
            axes[3].legend()
            axes[3].grid(True, alpha=0.3)
            
            plt.tight_layout()
            plt.show()
            
            print(f"   ✅ Charts created successfully for {symbol}")
            
        except Exception as e:
            print(f"   ❌ Chart creation failed for {symbol}: {e}")

print("\n🎉 Visualization testing complete!")
```
