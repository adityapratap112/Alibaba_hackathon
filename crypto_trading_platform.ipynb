## 4. Paper Trading Engine

```python
def execute_paper_trade(action, symbol, amount_usd, current_price):
    """Execute a paper trade (buy/sell)"""
    global user_portfolio
    
    if action.lower() == 'buy':
        if user_portfolio['balance_usd'] >= amount_usd:
            quantity = amount_usd / current_price
            user_portfolio['balance_usd'] -= amount_usd
            
            if symbol in user_portfolio['holdings']:
                user_portfolio['holdings'][symbol] += quantity
            else:
                user_portfolio['holdings'][symbol] = quantity
            
            # Record trade
            trade = {
                'timestamp': datetime.now(),
                'action': 'BUY',
                'symbol': symbol,
                'quantity': quantity,
                'price': current_price,
                'amount_usd': amount_usd
            }
            user_portfolio['trade_history'].append(trade)
            
            return f"✅ Bought {quantity:.6f} {symbol} for ${amount_usd:.2f}"
        else:
            return "❌ Insufficient balance"
    
    elif action.lower() == 'sell':
        if symbol in user_portfolio['holdings']:
            quantity_to_sell = amount_usd / current_price
            if user_portfolio['holdings'][symbol] >= quantity_to_sell:
                user_portfolio['holdings'][symbol] -= quantity_to_sell
                user_portfolio['balance_usd'] += amount_usd
                
                # Record trade
                trade = {
                    'timestamp': datetime.now(),
                    'action': 'SELL',
                    'symbol': symbol,
                    'quantity': quantity_to_sell,
                    'price': current_price,
                    'amount_usd': amount_usd
                }
                user_portfolio['trade_history'].append(trade)
                
                return f"✅ Sold {quantity_to_sell:.6f} {symbol} for ${amount_usd:.2f}"
            else:
                return "❌ Insufficient holdings"
        else:
            return f"❌ No {symbol} holdings to sell"

def display_portfolio():
    """Display current portfolio status"""
    print("\n💼 CURRENT PORTFOLIO")
    print("=" * 40)
    print(f"💰 USD Balance: ${user_portfolio['balance_usd']:.2f}")
    print("\n🪙 Cryptocurrency Holdings:")
    
    if user_portfolio['holdings']:
        for symbol, quantity in user_portfolio['holdings'].items():
            if quantity > 0:
                print(f"   {symbol}: {quantity:.6f}")
    else:
        print("   No cryptocurrency holdings")
    
    print(f"\n📊 Total Trades: {len(user_portfolio['trade_history'])}")

print("🔄 Paper trading engine ready!")
```
